```python
class TimeEmbed(nn.Module):
    def __init__(self, dim):
        super().__init__()
        self.mlp = nn.Sequential(
            nn.Linear(1, dim),
            nn.SiLU(),
            nn.Linear(dim, dim)
        )

    def forward(self, t):
        return self.mlp(t[:, None])

class EncDec(nn.Module):
    def __init__(self, in_ch=3, latent_ch=64, base=32):
        super().__init__()
        self.time = TimeEmbed(latent_ch)
        self.enc = nn.Sequential(
            nn.Conv2d(in_ch, base, 4, stride=2, padding=1),      # 32->16
            nn.ReLU(),
            nn.Conv2d(base, base, 3, stride=1, padding=1),       # 16->16
            nn.ReLU(),
            nn.Conv2d(base, base, 3, stride=2, padding=1),       # 16->6
            nn.ReLU(),
            nn.Conv2d(base, latent_ch, 3, stride=1, padding=1),  # 8->8
        )
        self.dec = nn.Sequential(
            nn.ConvTranspose2d(latent_ch, base, 3, stride=2, padding=0),
            nn.ReLU(),
            nn.ConvTranspose2d(base, base, 3, stride=2, padding=1),
            nn.ReLU(),
            nn.ConvTranspose2d(base, in_ch, 2, stride=1, padding=1),
        )
    
    def forward(self, x, t):
        z = self.enc(x)
        t_emb = self.time(t)
        z = z + t_emb[:, :, None, None]
        x_hat = self.dec(z)
        return x_hat, z

class Block(nn.Module):
    def __init__(self, in_ch, out_ch, t_dim):
        super().__init__()
        self.conv1 = nn.Conv2d(in_ch, out_ch, 3, padding=1)
        self.conv2 = nn.Conv2d(out_ch, out_ch, 3, padding=1)
        self.time_emb = nn.Linear(t_dim, out_ch)

    def forward(self, x, t):
        h = F.silu(self.conv1(x))
        h = h + self.time_emb(t)[:, :, None, None]
        h = F.silu(self.conv2(h))
        return h

class StandardClassifier(nn.Module):
    def __init__(self):
        super().__init__()
        base = 32
        t_dim = 32

        self.time_embed = TimeEmbed(t_dim)

        self.down1 = Block(3, base, t_dim)
        self.down2 = Block(base, base * 2, t_dim)
        self.down3 = Block(base * 2, base * 2, t_dim)
        self.pool = nn.MaxPool2d(2)

        self.mid = Block(base * 2, base * 2, t_dim)

        self.out = nn.Sequential(
            nn.AdaptiveAvgPool2d(1),
            nn.Flatten(),
            nn.Linear(base * 2, 2)
        )
        
    def forward(self, x, t):
        t = self.time_embed(t)

        d1 = self.down1(x, t)
        d2 = self.down2(self.pool(d1), t)
        d3 = self.down3(self.pool(d2), t)

        m = self.mid(d3, t)

        return self.out(m)


import math
import torch
from torch import nn
from torch.nn import init
from torch.nn import functional as F


class Swish(nn.Module):
    def forward(self, x):
        return x * torch.sigmoid(x)


class TimeEmbedding(nn.Module):
    def __init__(self, T, d_model, dim):
        assert d_model % 2 == 0
        super().__init__()
        emb = torch.arange(0, d_model, step=2) / d_model * math.log(10000)
        emb = torch.exp(-emb)
        pos = torch.arange(T).float()
        emb = pos[:, None] * emb[None, :]
        assert list(emb.shape) == [T, d_model // 2]
        emb = torch.stack([torch.sin(emb), torch.cos(emb)], dim=-1)
        assert list(emb.shape) == [T, d_model // 2, 2]
        emb = emb.view(T, d_model)

        self.timembedding = nn.Sequential(
            nn.Embedding.from_pretrained(emb),
            nn.Linear(d_model, dim),
            Swish(),
            nn.Linear(dim, dim),
        )
        self.initialize()

    def initialize(self):
        for module in self.modules():
            if isinstance(module, nn.Linear):
                init.xavier_uniform_(module.weight)
                init.zeros_(module.bias)

    def forward(self, t):
        emb = self.timembedding(t)
        return emb


class DownSample(nn.Module):
    def __init__(self, in_ch):
        super().__init__()
        self.main = nn.Conv2d(in_ch, in_ch, 3, stride=2, padding=1)
        self.initialize()

    def initialize(self):
        init.xavier_uniform_(self.main.weight)
        init.zeros_(self.main.bias)

    def forward(self, x, temb):
        x = self.main(x)
        return x


class UpSample(nn.Module):
    def __init__(self, in_ch):
        super().__init__()
        self.main = nn.Conv2d(in_ch, in_ch, 3, stride=1, padding=1)
        self.initialize()

    def initialize(self):
        init.xavier_uniform_(self.main.weight)
        init.zeros_(self.main.bias)

    def forward(self, x, temb):
        _, _, H, W = x.shape
        x = F.interpolate(
            x, scale_factor=2, mode='nearest')
        x = self.main(x)
        return x


class AttnBlock(nn.Module):
    def __init__(self, in_ch):
        super().__init__()
        self.group_norm = nn.GroupNorm(32, in_ch)
        self.proj_q = nn.Conv2d(in_ch, in_ch, 1, stride=1, padding=0)
        self.proj_k = nn.Conv2d(in_ch, in_ch, 1, stride=1, padding=0)
        self.proj_v = nn.Conv2d(in_ch, in_ch, 1, stride=1, padding=0)
        self.proj = nn.Conv2d(in_ch, in_ch, 1, stride=1, padding=0)
        self.initialize()

    def initialize(self):
        for module in [self.proj_q, self.proj_k, self.proj_v, self.proj]:
            init.xavier_uniform_(module.weight)
            init.zeros_(module.bias)
        init.xavier_uniform_(self.proj.weight, gain=1e-5)

    def forward(self, x):
        B, C, H, W = x.shape
        h = self.group_norm(x)
        q = self.proj_q(h)
        k = self.proj_k(h)
        v = self.proj_v(h)

        q = q.permute(0, 2, 3, 1).view(B, H * W, C)
        k = k.view(B, C, H * W)
        w = torch.bmm(q, k) * (int(C) ** (-0.5))
        assert list(w.shape) == [B, H * W, H * W]
        w = F.softmax(w, dim=-1)

        v = v.permute(0, 2, 3, 1).view(B, H * W, C)
        h = torch.bmm(w, v)
        assert list(h.shape) == [B, H * W, C]
        h = h.view(B, H, W, C).permute(0, 3, 1, 2)
        h = self.proj(h)

        return x + h


class ResBlock(nn.Module):
    def __init__(self, in_ch, out_ch, tdim, dropout, attn=False):
        super().__init__()
        self.block1 = nn.Sequential(
            nn.GroupNorm(32, in_ch),
            Swish(),
            nn.Conv2d(in_ch, out_ch, 3, stride=1, padding=1),
        )
        self.temb_proj = nn.Sequential(
            Swish(),
            nn.Linear(tdim, out_ch),
        )
        self.block2 = nn.Sequential(
            nn.GroupNorm(32, out_ch),
            Swish(),
            nn.Dropout(dropout),
            nn.Conv2d(out_ch, out_ch, 3, stride=1, padding=1),
        )
        if in_ch != out_ch:
            self.shortcut = nn.Conv2d(in_ch, out_ch, 1, stride=1, padding=0)
        else:
            self.shortcut = nn.Identity()
        if attn:
            self.attn = AttnBlock(out_ch)
        else:
            self.attn = nn.Identity()
        self.initialize()

    def initialize(self):
        for module in self.modules():
            if isinstance(module, (nn.Conv2d, nn.Linear)):
                init.xavier_uniform_(module.weight)
                init.zeros_(module.bias)
        init.xavier_uniform_(self.block2[-1].weight, gain=1e-5)

    def forward(self, x, temb):
        h = self.block1(x)
        h += self.temb_proj(temb)[:, :, None, None]
        h = self.block2(h)

        h = h + self.shortcut(x)
        h = self.attn(h)
        return h


class UNet(nn.Module):
    def __init__(self, T, ch, ch_mult, attn, num_res_blocks, dropout):
        super().__init__()
        assert all([i < len(ch_mult) for i in attn]), 'attn index out of bound'
        tdim = ch * 4
        self.time_embedding = TimeEmbedding(T, ch, tdim)

        self.head = nn.Conv2d(3, ch, kernel_size=3, stride=1, padding=1)
        self.downblocks = nn.ModuleList()
        chs = [ch]  # record output channel when dowmsample for upsample
        now_ch = ch
        for i, mult in enumerate(ch_mult):
            out_ch = ch * mult
            for _ in range(num_res_blocks):
                self.downblocks.append(ResBlock(
                    in_ch=now_ch, out_ch=out_ch, tdim=tdim,
                    dropout=dropout, attn=(i in attn)))
                now_ch = out_ch
                chs.append(now_ch)
            if i != len(ch_mult) - 1:
                self.downblocks.append(DownSample(now_ch))
                chs.append(now_ch)

        self.middleblocks = nn.ModuleList([
            ResBlock(now_ch, now_ch, tdim, dropout, attn=True),
            ResBlock(now_ch, now_ch, tdim, dropout, attn=False),
        ])

        self.upblocks = nn.ModuleList()
        for i, mult in reversed(list(enumerate(ch_mult))):
            out_ch = ch * mult
            for _ in range(num_res_blocks + 1):
                self.upblocks.append(ResBlock(
                    in_ch=chs.pop() + now_ch, out_ch=out_ch, tdim=tdim,
                    dropout=dropout, attn=(i in attn)))
                now_ch = out_ch
            if i != 0:
                self.upblocks.append(UpSample(now_ch))
        assert len(chs) == 0

        self.tail = nn.Sequential(
            nn.GroupNorm(32, now_ch),
            Swish(),
            nn.Conv2d(now_ch, 3, 3, stride=1, padding=1)
        )
        self.initialize()

    def initialize(self):
        init.xavier_uniform_(self.head.weight)
        init.zeros_(self.head.bias)
        init.xavier_uniform_(self.tail[-1].weight, gain=1e-5)
        init.zeros_(self.tail[-1].bias)

    def forward(self, x, t):
        # Timestep embedding
        temb = self.time_embedding(t)
        # Downsampling
        h = self.head(x)
        hs = [h]
        for layer in self.downblocks:
            h = layer(h, temb)
            hs.append(h)
        # Middle
        for layer in self.middleblocks:
            h = layer(h, temb)
        # Upsampling
        for layer in self.upblocks:
            if isinstance(layer, ResBlock):
                h = torch.cat([h, hs.pop()], dim=1)
            h = layer(h, temb)
        h = self.tail(h)

        assert len(hs) == 0
        return h


class Classifier(nn.Module):
    def __init__(self):
        super().__init__()

        self.features = nn.Sequential(
            nn.Conv2d(3, 64, 3, padding=1),
            nn.Conv2d(64, 64, 3, padding=1),
            nn.GroupNorm(16, 64),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Dropout2d(0.1),

            nn.Conv2d(64, 128, 3, padding=1),
            nn.Conv2d(128, 128, 3, padding=1),
            nn.GroupNorm(16, 128),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Dropout2d(0.2),

            nn.Conv2d(128, 256, 3, padding=1),
            nn.Conv2d(256, 256, 3, padding=1),
            nn.GroupNorm(16, 256),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Dropout2d(0.3),

            nn.AdaptiveAvgPool2d(1),
            nn.Flatten()
        )

        self.fc = nn.Linear(256, 10)

    def forward(self, x):
        x = self.features(x)
        return self.fc(x)
```

```python
class SinusoidalPositionEmbeddings(nn.Module):
    """
    Generates sinusoidal embeddings for timesteps (Ho et al., 2020).
    """
    def __init__(self, dim):
        super().__init__()
        self.dim = dim

    def forward(self, time):
        device = time.device
        half_dim = self.dim // 2
        embeddings = math.log(10000) / (half_dim - 1)
        embeddings = torch.exp(torch.arange(half_dim, device=device) * -embeddings)
        embeddings = time[:, None] * embeddings[None, :]
        embeddings = torch.cat((embeddings.sin(), embeddings.cos()), dim=-1)
        return embeddings

class ConvAutoencoder(nn.Module):
    def __init__(self, latent_dim=128, time_emb_dim=128):
        super().__init__()

        self.time_mlp = nn.Sequential(
            SinusoidalPositionEmbeddings(time_emb_dim),
            nn.Linear(time_emb_dim, time_emb_dim),
            nn.GELU(),
            nn.Linear(time_emb_dim, time_emb_dim)
        )

        self.time_emb_enc1 = nn.Linear(time_emb_dim, 32)
        self.time_emb_enc2 = nn.Linear(time_emb_dim, 64)
        self.time_emb_enc3 = nn.Linear(time_emb_dim, 128)

        self.time_emb_dec1 = nn.Linear(time_emb_dim, 128)
        self.time_emb_dec2 = nn.Linear(time_emb_dim, 64)
        self.time_emb_dec3 = nn.Linear(time_emb_dim, 32)

        self.enc_conv1 = nn.Conv2d(3, 32, 3, stride=2, padding=1)  # B, 32, 16, 16
        self.enc_conv2 = nn.Conv2d(32, 64, 3, stride=2, padding=1) # B, 64, 8, 8
        self.enc_conv3 = nn.Conv2d(64, 128, 3, stride=2, padding=1)# B, 128, 4, 4

        self.flatten = nn.Flatten()
        self.fc_enc = nn.Linear(128 * 4 * 4, latent_dim)

        self.fc_dec = nn.Linear(latent_dim, 128 * 4 * 4)

        self.dec_conv1 = nn.ConvTranspose2d(128, 64, 3, stride=2, padding=1, output_padding=1) # 8x8
        self.dec_conv2 = nn.ConvTranspose2d(64, 32, 3, stride=2, padding=1, output_padding=1)  # 16x16
        self.dec_conv3 = nn.ConvTranspose2d(32, 3, 3, stride=2, padding=1, output_padding=1)   # 32x32

        self.relu = nn.ReLU()
        self.tanh = nn.Tanh()

    def encode(self, x, t):
        t_emb = self.time_mlp(t)

        x = self.enc_conv1(x)
        x = x + self.time_emb_enc1(t_emb).view(-1, 32, 1, 1)
        x = self.relu(x)

        x = self.enc_conv2(x)
        x = x + self.time_emb_enc2(t_emb).view(-1, 64, 1, 1)
        x = self.relu(x)

        x = self.enc_conv3(x)
        x = x + self.time_emb_enc3(t_emb).view(-1, 128, 1, 1)
        x = self.relu(x)

        x = self.flatten(x)
        z = self.fc_enc(x)
        return z, t_emb

    def decode(self, z, t_emb):
        x = self.fc_dec(z)
        x = x.view(-1, 128, 4, 4)

        x = x + self.time_emb_dec1(t_emb).view(-1, 128, 1, 1)
        x = self.dec_conv1(x)
        x = self.relu(x)

        x = x + self.time_emb_dec2(t_emb).view(-1, 64, 1, 1)
        x = self.dec_conv2(x)
        x = self.relu(x)

        x = x + self.time_emb_dec3(t_emb).view(-1, 32, 1, 1)
        x = self.dec_conv3(x)
        x = self.tanh(x)

        return x

    def forward(self, x, t):
        z, t_emb = self.encode(x, t)
        out = self.decode(z, t_emb)
        return out
```

```python
# ---- Your starting lines ----
from datasets import load_dataset
from torch.utils.data import DataLoader
import torch
from torchvision import transforms
dataset = load_dataset("cifar10")

import numpy as np

transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
])

def hf_split_to_torchstyle(hf_split):

    cols = [c.lower() for c in hf_split.column_names]
    if "image" in cols:
        img_col = hf_split.column_names[cols.index("image")]
    elif "img" in cols:
        img_col = hf_split.column_names[cols.index("img")]
    else:
        raise ValueError(f"No image column found. Columns: {hf_split.column_names}")

    images = []
    labels = []

    for item in hf_split:
        img = item[img_col]

        # Convert PIL → numpy if needed
        if hasattr(img, "convert") or hasattr(img, "mode"):
            arr = np.array(img)
        else:
            arr = np.asarray(img)

        images.append(arr)
        labels.append(item["label"])

    images = np.stack(images)   # (N, 32, 32, 3)
    labels = np.array(labels)

    class TorchStyle:
        pass

    t = TorchStyle()
    t.data = images
    t.targets = labels
    return t

train_set = hf_split_to_torchstyle(dataset["train"])
test_set  = hf_split_to_torchstyle(dataset["test"])
```

```python
animal_labels = {2, 3, 4, 5, 6, 7}
flat_dim = 32 * 32 * 3  # 3072

X = torch.from_numpy(train_set.data).float() / 255.0

targets_np = np.array(train_set.targets)
mask = np.isin(targets_np, list(animal_labels))

Y = torch.from_numpy(train_set.data[mask]).cpu().float() / 255.0

Y = (Y - 0.5) / 0.5
X = (X - 0.5) / 0.5

print("X.shape =", X.shape)
print("Y.shape =", Y.shape)
```

```python
img_size = 32
batch_size = 128
epochs = 100
diffusion_steps = timesteps = 1000

beta = torch.linspace(1e-4, 0.02, diffusion_steps)
alpha = 1. - beta
baralphas = alpha_bar = torch.cumprod(alpha, dim=0)
```

```python
def sample_timesteps(batch_size, diffusion_steps, power=2.0):
    u = torch.rand(batch_size, 1)
    t = (u ** power * diffusion_steps).long()
    return t

def noise(Xbatch, t):
    eps = torch.randn_like(Xbatch)

    a = baralphas[t].sqrt().view(-1, 1, 1, 1)
    b = (1 - baralphas[t]).sqrt().view(-1, 1, 1, 1)

    noised = a * Xbatch + b * eps
    return noised, eps


diffusion_steps = 1000

X_rand_timesteps = sample_timesteps(X.shape[0], diffusion_steps)
X, _ = noise(X, X_rand_timesteps)

Y_rand_timesteps = sample_timesteps(Y.shape[0], diffusion_steps)
Y, _ = noise(Y, Y_rand_timesteps)
```

```python
with torch.no_grad():
    phi_all, _ = encoder.encode(X.permute(0,3,1,2).cuda(), X_rand_timesteps.cuda())
    phi_eight, _ = encoder.encode(Y.permute(0,3,1,2).cuda(), Y_rand_timesteps.cuda())


n_de = phi_all.shape[0]
n_nu = phi_eight.shape[0]
reg = 0.01

H = (phi_all.T @ phi_all) / n_de
H += reg * torch.eye(128).to("cuda")

h = torch.mean(phi_eight, dim=0, keepdim=True).T

theta = torch.linalg.solve(H, h)
```
