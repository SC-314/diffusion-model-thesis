```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

def draw_block(ax, xy, width, height, text):
    rect = Rectangle(xy, width, height, fill=False)
    ax.add_patch(rect)
    ax.text(
        xy[0] + width / 2,
        xy[1] + height / 2,
        text,
        ha='center',
        va='center'
    )

def draw_arrow(ax, start, end):
    ax.annotate(
        "",
        xy=end,
        xytext=start,
        arrowprops=dict(arrowstyle="->")
    )

fig, ax = plt.subplots()

# Block positions
block_width = 2
block_height = 1

input_pos = (0, 2)
encoder_pos = (3, 2)
bottleneck_pos = (6, 2)
decoder_pos = (9, 2)
output_pos = (12, 2)

# Draw blocks
draw_block(ax, input_pos, block_width, block_height, "Input\n(784)")
draw_block(ax, encoder_pos, block_width, block_height, "Encoder\n(128)")
draw_block(ax, bottleneck_pos, block_width, block_height, "Latent\n(32)")
draw_block(ax, decoder_pos, block_width, block_height, "Decoder\n(128)")
draw_block(ax, output_pos, block_width, block_height, "Output\n(784)")

# Draw arrows
draw_arrow(ax, (2, 2.5), (3, 2.5))
draw_arrow(ax, (5, 2.5), (6, 2.5))
draw_arrow(ax, (8, 2.5), (9, 2.5))
draw_arrow(ax, (11, 2.5), (12, 2.5))

ax.set_xlim(-1, 15)
ax.set_ylim(0, 5)
ax.set_aspect('equal')
ax.axis('off')

plt.title("Autoencoder Architecture")
plt.show()
```
